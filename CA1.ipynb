## **Step 4** - Data Cleaning (Preprocessing)
Having identified certain nulls in the previous section, cleaning is required.

```python
# POI
poi_df = poi_df.dropna()
poi_df.isnull().sum()
poi_df = poi_df[poi_df['POI_Type'] != 'Other']
poi_df['POI_Type'] = poi_df['POI_Type'].str.strip().str.lower().str.replace('busstop', 'bus stop')
poi_df['POI_Type'] = poi_df['POI_Type'].str.title()
poi_df['POI_Type'].value_counts()
```

```python
# Private Property
private_property_df = private_property_df.fillna(0)
private_property_df = private_property_df.drop_duplicates()

```

```python
# Sales
cols_to_fill = ['PropertyNumberOfFloors', 'PropertynumberOfUnits', 'PropertyYearOfCompletion', 'NumberOfBedroomsInTransactedUnit']

for col in cols_to_fill:
    sales_df[col] = sales_df[col].fillna(sales_df[col].median())

sales_df = sales_df.drop_duplicates() 
```

## **Step 5** Feature Engineering
Transforming and manipulating data to create new columns that relates to the main objective to understand it better.

```python
# Property Space
property_space_df['TimeToRafflesInMins_Car'] = round(property_space_df['TimeToRafflesByCarInSeconds'] / 60, 2)
property_space_df['TimeToRafflesInMins_PT'] = round(property_space_df['TimeToRafflesByPublicTransportInSeconds'] / 60, 2)

property_space_df['TimeToAirportInMins_Car'] = round(property_space_df['TimeToAirportByCarInSeconds'] / 60, 2)
property_space_df['TimeToAirportInMins_PT'] = round(property_space_df['TimeToAirportByPublicTransportInSeconds'] / 60, 2)


property_space_df['CityProximityCategory_PT'] = pd.cut(property_space_df['TimeToRafflesInMins_PT'],
                                                      bins=[0, 15, 30, float('inf')],
                                                      labels=['<15 min', '15-30 min', '30+ min'])

property_space_df['CityProximityCategory_Car'] = pd.cut(property_space_df['TimeToRafflesInMins_Car'],
                                                      bins=[0, 15, 30, float('inf')],
                                                      labels=['<15 min', '15-30 min', '30+ min'])

property_space_df['AccessibilityScore'] = (property_space_df['NumberofSchoolsWithin500Mts'] + 
                                            property_space_df['NumberOfMRTsWithin500Mts'] + 
                                            property_space_df['NumberOfBusStopsWithin500Mts'])


# Sales
def floor_category(floor):
    if floor < 0:
        return 'Basement'
    elif floor == 0:
        return 'Ground Floor'
    elif floor <= 10:
        return 'Low'
    elif floor <= 20:
        return 'Mid'
    else:
        return 'High'
    
sales_df['FloorCategory'] = sales_df['FloorNoOfTransactedUnit'].apply(floor_category)

type_labels = {
    1: 'Apartment',
    2: 'Executive Condominium',
    3: 'Detached House',
    4: 'HDB',
    5: 'Semi Detached House',
    6: 'Terraced House',
    7: 'Condominium'
}

type_categories = {
    1: 'Private',
    2: 'Private',
    3: 'Landed',
    4: 'Public',
    5: 'Landed',
    6: 'Landed',
    7: 'Private'
}

sales_df['PropertyTypeLabel'] = sales_df['PropertyType'].map(type_labels)
sales_df['PropertyCategory'] = sales_df['PropertyType'].map(type_categories)

transaction_labels = {
    0: 'NewSale',
    1: 'Subsale',
    2: 'Resale'
}

sales_df['TransactionType'] = sales_df['TransactionType'].map(transaction_labels)

tenure_labels = {
    1: 'Freehold', 
    2: 'Leasehold (99 Years)',
    3: 'Unknown'
}

sales_df['PropertyTenureType'] = sales_df['PropertyTenureType'].map(tenure_labels)



sales_df['BedroomCategory'] = pd.cut(sales_df['NumberOfBedroomsInTransactedUnit'],
                                      bins=[0, 1, 2, 3, float('inf')],
                                      labels=['1', '2', '3', '4+'])

# Private Property

amenity_cols = ['Gym', 'Pool', 'Playground', 'Tennis', 'Golf', 'Cafe', 'LapPool', 'KidsPool', 'Squash', 'Badminton', 'Basketball', 'BBQ', 'Jacuzzi', 'Security', 'Salon', 'Sauna', 'Shop']
private_property_df[amenity_cols] = private_property_df[amenity_cols].apply(pd.to_numeric, errors='coerce').fillna(0)
private_property_df['AmenityCount'] = private_property_df[amenity_cols].sum(axis=1)

```
